# EDA 05: Geographic Analysis

Analyzing country and location risk.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

tx_df = pd.read_csv('../../data/original/transactions.csv')
ml_df = pd.read_csv('../../data/original/ml_features.csv')
df = tx_df.copy()
df['is_suspicious_tx'] = ml_df['is_suspicious_tx']
```

## Country Distribution

```
if 'Sender_bank_location' in df.columns:
    plt.figure(figsize=(10, 5))
    df['Sender_bank_location'].value_counts().plot(kind='bar')
    plt.title('Sender Bank Location')
    plt.show()

if 'Receiver_bank_location' in df.columns:
    plt.figure(figsize=(10, 5))
    df['Receiver_bank_location'].value_counts().plot(kind='bar')
    plt.title('Receiver Bank Location')
    plt.show()
```

## Cross Border Transactions

```
if 'Sender_bank_location' in df.columns and 'Receiver_bank_location' in df.columns:
    df['is_cross_border'] = (df['Sender_bank_location'] != df['Receiver_bank_location']).astype(int)
    print('Cross Border Ratio:')
    print(df['is_cross_border'].value_counts(normalize=True))
```

## Fraud Rate by Country

```
if 'Sender_bank_location' in df.columns:
    fraud_by_sender_country = df.groupby('Sender_bank_location')['is_suspicious_tx'].mean().sort_values(ascending=False)
    fraud_by_sender_country.plot(kind='bar', figsize=(10, 5))
    plt.title('Suspicious Rate by Sender Country')
    plt.show()
```
